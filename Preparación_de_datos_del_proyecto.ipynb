1.   **Lectura de los datos en Python.**

Leemos las imágenes y guardamos en un arreglo, ajustamos tamaño y visualizamos.

Utilizamos las librerías de:
*   **cv2:** para el reconocimiento de imágenes.
*   **numpy:** para los arreglos.
*   **os:** para los directorios.
*   **matplolib:** para ver las gráficas.


```python
ruta_enfermas = "/content/Enfermas"
enfermas=[]
img_size=150
```

Se crea una ruta y definimos el tamaño de las imágenes.

```python
for img in os.listdir(ruta_enfermas):
  img = cv2.imread(os.path.join(ruta_enfermas, img))
  img_resize = cv2.resize(img,(img_size,img_size))
  enfermas.append(img_resize)
```

Se realiza un ciclo for para redimensionar al mismo tamaño todas las imágenes.

```python
print(len(enfermas))
print(enfermas)
```

Imprimimos el arreglo.

```python
enfermas = np.array(enfermas)
print(enfermas.shape)
```

Revisamos la información de las imágenes (número de imágenes, tamaño de imagen alto, tamaño imagen largo, canales de color)
